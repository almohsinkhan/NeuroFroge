```python
# import pytorch , sklearn , matplotlin and pillow ( for image handling)
import torch
import torchvision
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
```

```python
import torchvision
import torchvision.transforms as transforms

dataset = torchvision.datasets.FashionMNIST(
    root="./data",
    train=True,
    download=True,
    transform=transforms.ToTensor()
)

dataset_size = len(dataset)

print("Dataset size:", dataset_size)
```

```python
# look at image shape and label
image, label = dataset[0]
print("Image shape:", image.shape)
print("Label:", label)
```

```python
import numpy as np
import os

images = dataset.data.numpy()
labels = dataset.targets.numpy()

# Normalize: 0-255 → 0-1
images = images / 255.0

# Flatten: 28x28 → 784
images = images.reshape(60000, 784)

# Use only 1000 samples for now
images = images[:1000]
labels = labels[:1000]

os.makedirs("./data", exist_ok=True)

np.savetxt("./data/images.csv", images, delimiter=",")
np.savetxt("./data/labels.csv", labels, delimiter=",", fmt="%d")

print("Images:", images.shape)
print("Labels:", labels.shape)
```
